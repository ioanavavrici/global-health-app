# Lab 2 - Data Exploration and Missing Values (Task 2.1)

Acest notebook apeleaz? extractorii pentru **World Bank**, **WHO** ?i **Our World in Data**, transform? fi?ierele brute ?n formatul normalizat (`country_iso3`, `indicator_code`, `year`, `sex`, `value`) ?i inspecteaz? structura ?i ponderea valorilor lips? pentru fiecare indicator.

```python
import os
from pathlib import Path
import pandas as pd
from IPython.display import display

# Asigur? rularea din r?d?cina proiectului indiferent de unde e deschis notebook-ul
if Path.cwd().name == 'notebooks':
    os.chdir('..')

from etl.extract_worldbank import download_worldbank_countries, download_worldbank_indicator
from etl.extract_who import download_who_indicator
from etl.extract_owid import download_owid_chart
from etl.transform_worldbank import transform_worldbank_file
from etl.transform_who import transform_who_file
from etl.transform_owid import transform_owid_file
```

```python
# 1. Extragere ?i transformare World Bank
download_worldbank_countries()
wb_codes = ['SH.XPD.CHEX.GD.ZS', 'SH.IMM.MEAS', 'SP.DYN.IMRT.IN', 'NY.GDP.PCAP.CD']
dfs = {}

for code in wb_codes:
    path = download_worldbank_indicator(code)
    df = transform_worldbank_file(path, code)
    dfs[code] = df
    print(f'=== World Bank: {code} ===')
    display(df.head())
    df.info()
    print('Missing values:\n', df.isna().sum())
    print()
```

```python
# 2. Extragere ?i transformare WHO (WHOSIS_000001)
who_path, _ = download_who_indicator('WHOSIS_000001')
df_who = transform_who_file(who_path, 'WHOSIS_000001')
dfs['WHOSIS_000001'] = df_who
print('=== WHO: WHOSIS_000001 ===')
display(df_who.head())
df_who.info()
print('Missing values:\n', df_who.isna().sum())
```

```python
# 3. Extragere ?i transformare Our World in Data (life-expectancy)
owid_path, _ = download_owid_chart('life-expectancy')
df_owid = transform_owid_file(owid_path, 'life-expectancy')
dfs['life-expectancy'] = df_owid
print('=== OWID: life-expectancy ===')
display(df_owid.head())
df_owid.info()
print('Missing values:\n', df_owid.isna().sum())
```

```python
# 4. Sumar pondere valori lips? (%) per indicator
summary = []
for code, df in dfs.items():
    total = len(df)
    missing_val = int(df['value'].isna().sum())
    missing_iso = int(df['country_iso3'].isna().sum())
    share_pct = round(100 * missing_val / total, 2) if total else 0.0
    summary.append({
        'indicator': code,
        'raw_rows': total,
        'missing_value': missing_val,
        'missing_value_pct': share_pct,
        'missing_country_iso3': missing_iso,
    })
summary_df = pd.DataFrame(summary)
display(summary_df)
print(summary_df.to_string(index=False))
```
