# Lab 2 - Data Exploration and Missing Values
This notebook loads the extracted raw data, transforms it, and inspects missing values and data shapes per indicator.

```python
import pandas as pd
from pathlib import Path
from etl.transform_worldbank import transform_worldbank_file
from etl.transform_who import transform_who_file
from etl.transform_owid import transform_owid_file

# 1. Inspect World Bank Maternal Mortality
wb_path = Path('data/raw/worldbank/SH.STA.MMRT_from2000.json')
if wb_path.exists():
    df_wb = transform_worldbank_file(wb_path, 'SH.STA.MMRT')
    display(df_wb.head())
    print(df_wb.info())
    print(df_wb.isna().sum())
```

```python
# 2. Inspect WHO Skilled Birth Attendance
who_path = Path('data/raw/who/MDG_0000000025_from2000.json')
if who_path.exists():
    df_who = transform_who_file(who_path, 'MDG_0000000025')
    display(df_who.head())
    print(df_who.info())
    print(df_who.isna().sum())
```

```python
# 3. Inspect OWID Life Expectancy
owid_path = Path('data/raw/owid/life-expectancy.csv')
if owid_path.exists():
    df_owid = transform_owid_file(owid_path, 'life-expectancy')
    display(df_owid.head())
    print(df_owid.info())
    print(df_owid.isna().sum())
```
